```python
import numpy as np
import pandas as pd
from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import r2_score

# Load data
df = pd.read_csv('https://www.dropbox.com/scl/fi/p7nbfatqu7ygrqess0oq0/Cars-detail-for-model.csv?rlkey=l8amm4dbgspt1hq3coom0nben&st=vgis5hhl&dl=1')

# Data preprocessing
ln_fuel = LabelEncoder()
df['fuel'] = ln_fuel.fit_transform(df['fuel'])
ln_seller_type = LabelEncoder()
df['seller_type'] = ln_seller_type.fit_transform(df['seller_type'])
ln_owner = LabelEncoder()
df['owner'] = ln_owner.fit_transform(df['owner'])
ln_transmission = LabelEncoder()
df['transmission'] = ln_transmission.fit_transform(df['transmission'])

x = df.drop(columns = ['name', 'selling_price'] )
y = df['selling_price']

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size = 0.2, random_state = 2)

# Train the model
model = RandomForestRegressor(n_estimators = 100, random_state = 2)
model.fit(x_train, y_train)

#Test Examples
# Format: [year, km_driven, fuel, seller_type, transmission, owner]
examples = [
    {"desc": "2017 Petrol", "data": [2017, 45000, 4, 1, 1, 0]},
    {"desc": "2012 Budget Petrol (Manual)", "data": [2012, 80000, 4, 1, 1, 2]},
    {"desc": "2019 Luxury Diesel (Automatic)", "data": [2019, 20000, 1, 0, 0, 0]},
    {"desc": "2008 Older Car (Petrol)", "data": [2008, 120000, 4, 1, 1, 4]},
    {"desc": "2016 CNG Economy", "data": [2016, 55000, 0, 1, 1, 0]}
]

# 2. Predict and print results
print(f"R-squared score: {r2_score(y_test, model.predict(x_test)):.2f}\n")
print("--- TEST PREDICTIONS ---")

for item in examples:
    test_df = pd.DataFrame([item["data"]], columns=x.columns)
    prediction = model.predict(test_df)[0]
    print(f"{item['desc']:30} | Predicted Price: {prediction:,.2f}")
```
